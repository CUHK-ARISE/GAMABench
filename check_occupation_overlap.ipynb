```python
import json
from itertools import combinations

# Load metadata.json
with open('public/metadata.json', 'r') as f:
    metadata = json.load(f)

# Get CABIN categories (occupation -> question numbers)
cabin_categories = metadata['CABIN']['categories']

# Create reverse mapping: question number -> occupation name
question_to_occupation = {}
for occupation, questions in cabin_categories.items():
    for q in questions:
        question_to_occupation[q] = occupation

print(f"Total occupations: {len(cabin_categories)}")
```

```python
# Load CABIN-en.json
with open('public/CABIN-en.json', 'r') as f:
    cabin_en = json.load(f)

# Separate Six Broad Band and Eight Broad Band
six_band = {}
eight_band = {}

for item in cabin_en:
    name = item['cat_name']
    questions = item['cat_questions']
    
    # Get unique occupations for this dimension
    occupations = set()
    for q in questions:
        occ = question_to_occupation.get(q)
        if occ:
            occupations.add(occ)
    
    if 'Six Broad Band' in name:
        six_band[name] = occupations
    elif 'Eight Broad Band' in name:
        eight_band[name] = occupations

print("Six Broad Band dimensions:")
for name, occs in six_band.items():
    print(f"  {name}: {len(occs)} occupations")

print("\nEight Broad Band dimensions:")
for name, occs in eight_band.items():
    print(f"  {name}: {len(occs)} occupations")
```

```python
# Check overlaps in Six Broad Band
print("=" * 60)
print("SIX BROAD BAND - Occupation Overlaps")
print("=" * 60)

six_names = list(six_band.keys())
for d1, d2 in combinations(six_names, 2):
    overlap = six_band[d1] & six_band[d2]
    if overlap:
        d1_short = d1.replace('Six Broad Band ', '')
        d2_short = d2.replace('Six Broad Band ', '')
        print(f"\n{d1_short} & {d2_short}:")
        print(f"  Overlapping occupations ({len(overlap)}): {sorted(overlap)}")
```

```python
# Check overlaps in Eight Broad Band
print("=" * 60)
print("EIGHT BROAD BAND - Occupation Overlaps")
print("=" * 60)

eight_names = list(eight_band.keys())
for d1, d2 in combinations(eight_names, 2):
    overlap = eight_band[d1] & eight_band[d2]
    if overlap:
        d1_short = d1.replace('Eight Broad Band ', '')
        d2_short = d2.replace('Eight Broad Band ', '')
        print(f"\n{d1_short} & {d2_short}:")
        print(f"  Overlapping occupations ({len(overlap)}): {sorted(overlap)}")
```

```python
# Summary: Show each dimension's occupations
print("=" * 60)
print("SUMMARY: All Occupations by Dimension")
print("=" * 60)

print("\n--- Six Broad Band ---")
for name, occs in six_band.items():
    short_name = name.replace('Six Broad Band ', '')
    print(f"\n{short_name}:")
    for occ in sorted(occs):
        print(f"  - {occ}")

print("\n--- Eight Broad Band ---")
for name, occs in eight_band.items():
    short_name = name.replace('Eight Broad Band ', '')
    print(f"\n{short_name}:")
    for occ in sorted(occs):
        print(f"  - {occ}")
```

```python
# Create overlap matrix for visualization
import pandas as pd

def create_overlap_matrix(band_dict, band_name):
    names = list(band_dict.keys())
    short_names = [n.replace(f'{band_name} ', '') for n in names]
    
    matrix = []
    for i, n1 in enumerate(names):
        row = []
        for j, n2 in enumerate(names):
            if i == j:
                row.append(len(band_dict[n1]))
            else:
                overlap = len(band_dict[n1] & band_dict[n2])
                row.append(overlap)
        matrix.append(row)
    
    df = pd.DataFrame(matrix, index=short_names, columns=short_names)
    return df

print("Six Broad Band Overlap Matrix (diagonal = total occupations):")
display(create_overlap_matrix(six_band, 'Six Broad Band'))

print("\nEight Broad Band Overlap Matrix (diagonal = total occupations):")
display(create_overlap_matrix(eight_band, 'Eight Broad Band'))
```
