```python
import json

# Load metadata.json to get occupation mapping
with open('public/metadata.json', 'r') as f:
    metadata = json.load(f)

# Get CABIN categories (occupation -> question numbers)
cabin_categories = metadata['CABIN']['categories']
print(f"Total occupations: {len(cabin_categories)}")
print(list(cabin_categories.keys()))
```

```python
# Create reverse mapping: question number -> occupation name
question_to_occupation = {}

for occupation, questions in cabin_categories.items():
    for q in questions:
        question_to_occupation[q] = occupation

# Verify mapping
print(f"Total questions mapped: {len(question_to_occupation)}")
print("\nSample mappings:")
for q in [1, 5, 41, 69, 97, 161]:
    print(f"  Question {q}: {question_to_occupation.get(q, 'N/A')}")
```

```python
# Load CABIN-en.json
with open('public/CABIN-en.json', 'r') as f:
    cabin_en = json.load(f)

print(f"Total categories in CABIN-en.json: {len(cabin_en)}")
for item in cabin_en:
    print(f"  - {item['cat_name']}: {len(item['cat_questions'])} questions")
```

```python
# Add occupation field to each category
for item in cabin_en:
    occupations = []
    for q in item['cat_questions']:
        occ = question_to_occupation.get(q)
        if occ and occ not in occupations:
            occupations.append(occ)
    item['occupation'] = occupations

# Display result
for item in cabin_en:
    print(f"\n{item['cat_name']}:")
    print(f"  Occupations: {item['occupation']}")
```

```python
# Save updated CABIN-en.json
with open('public/CABIN-en.json', 'w') as f:
    json.dump(cabin_en, f, indent=4)

print("Saved to public/CABIN-en.json")
```
